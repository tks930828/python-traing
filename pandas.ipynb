```python
df["摂取カロリー"] = df.loc[:, "摂取カロリー"].astype(np.float32)
```

```python
df["歩数/カロリー"] = (df.loc[:, "歩数"]/
                      df.loc[:,"摂取カロリー"])
df
```

```python
def exercise_judge(ex):
    if ex <= 3.0:
        return "Low"
    elif 3.0 < ex <= 6.0:
        return "Mid"
    else:
        return "High"
```

```python
df["運動指数"] = df.loc[:, "歩数/カロリー"].apply(
    exercise_judge)
df
```

```python
df.to_pickle("data/df_202411health.pickle")
```
