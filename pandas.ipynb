```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(
  np.arange(6).reshape((3, 2)), 
  columns=['A列', 'B列'], 
  index=['1行', '2行', '3行'])
df
```

```python
df['B列'] >= 3
```

```python
df[df['B列'] >= 3]
```

```python
df[df.loc[:, 'B列'] >= 3]
```

```python
df.query('B列 >= 3')
```
