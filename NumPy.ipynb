```python
c1 = np.array([0, 1, 2, 3, 4, 5])
c1
```

```python
#変形（reshape)
#↓の場合は2行、３列の２次元配列に変換する
c2 = c1.reshape((2 , 3))
c2
```

```python
#変換（ravel)
c3 = c2.ravel()
c3
```

```python
#copyを返す
c4 = c2.flatten()
c4
```
